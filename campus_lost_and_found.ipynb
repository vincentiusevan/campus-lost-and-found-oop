## Bagian 1: Class `Item` dan `User` (Nailla)
Bagian ini mendefinisikan class `Item` untuk menyimpan data barang dan class `User` untuk menyimpan identitas pelapor dengan menerapkan **Encapsulation** (`@property` dan validasi *setter*).

```python
# ==========================================
# CELL 2 - BAGIAN 1: Item & User (Nailla)
# ==========================================
class Item:
    """Store information about a lost or found item."""

    ALLOWED_CATEGORIES = (
        "Electronics",
        "Documents",
        "Personal",
        "Others"
    )

    def __init__(self, name, category, description):
        """Initialize an Item with name, category, and description."""
        self.name = name
        self.category = category
        self.description = description

    @property
    def name(self):
        """Return the item name."""
        return self._name

    @name.setter
    def name(self, value):
        """Validate that the item name is a non-empty string."""
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Item name cannot be empty.")
        self._name = value.strip()

    @property
    def category(self):
        """Return the item category."""
        return self._category

    @category.setter
    def category(self, value):
        """Validate that the item category is one of the allowed categories."""
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Category cannot be empty.")

        cleaned = value.strip().capitalize()
        if cleaned not in self.ALLOWED_CATEGORIES:
            raise ValueError(
                f"Invalid category '{value}'. Choose from {self.ALLOWED_CATEGORIES}."
            )
        self._category = cleaned

    @property
    def description(self):
        """Return the item description."""
        return self._description

    @description.setter
    def description(self, value):
        """Validate that the item description is a non-empty string."""
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Item description cannot be empty.")
        self._description = value.strip()

    def matches(self, keyword):
        """Check if the keyword matches the item's name or category (case-insensitive)."""
        if not isinstance(keyword, str) or not keyword.strip():
            return False
        query = keyword.strip().lower()
        return query in self._name.lower() or query in self._category.lower()

    def __str__(self):
        """Return a concise string summary of the item."""
        return f"{self._name} [{self._category}] - {self._description}"

class User:
    """Store information about a system user."""

    def __init__(self, user_id, name, contact):
        """Initialize a User with user_id (NIM/ID), name, and contact info."""
        self.user_id = user_id
        self.name = name
        self.contact = contact

    @property
    def user_id(self):
        """Return the user's ID / NIM."""
        return self._user_id

    @user_id.setter
    def user_id(self, value):
        """Validate that user_id is a non-empty string."""
        if not isinstance(value, str) or not value.strip():
            raise ValueError("User ID (NIM) cannot be empty.")
        self._user_id = value.strip()

    @property
    def name(self):
        """Return the user's name."""
        return self._name

    @name.setter
    def name(self, value):
        """Validate that the user's name is a non-empty string."""
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Reporter name cannot be empty.")
        self._name = value.strip()

    @property
    def contact(self):
        """Return the user's contact information."""
        return self._contact

    @contact.setter
    def contact(self, value):
        """Validate that the contact information is a non-empty string."""
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Contact information cannot be empty.")
        self._contact = value.strip()

    def __str__(self):
        """Return formatted reporter identity: 'Name (NIM) - Contact'."""
        return f"{self._name} ({self._user_id}) - {self._contact}"
```

## Bagian 2: Class `Report`, `LostReport`, dan `FoundReport` (Evan)
Bagian ini menerapkan **Abstraction** (`ABC` & `@abstractmethod`), **Inheritance** (`super()`), **Polymorphism** (`show_details()` & `get_report_type()`), serta **Composition/Association** (*has-a relationship* dengan `Item` dan `User`).

```python
# ==========================================================
# CELL 3 - BAGIAN 2: Report, LostReport, FoundReport (Evan)
# ==========================================================
from abc import ABC, abstractmethod


class Report(ABC):
    """Abstract base class representing a general lost or found report."""

    VALID_STATUSES = ("OPEN", "RESOLVED")

    def __init__(self, report_id, item, location, reporter=None, date="-"):
        """Initialize a report with ID, Item object, location, optional User, and date."""
        self.report_id = report_id
        self.item = item
        self.location = location
        self.reporter = reporter
        self.date = date
        self._status = "OPEN"

    @property
    def report_id(self):
        """Return the unique report ID."""
        return self._report_id

    @report_id.setter
    def report_id(self, value):
        """Validate that the report ID is a non-empty string."""
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Report ID cannot be empty.")
        self._report_id = value.strip().upper()

    @property
    def item(self):
        """Return the associated Item object."""
        return self._item

    @item.setter
    def item(self, value):
        """Validate that item is a valid Item object."""
        if value is None or not isinstance(value, Item):
            raise ValueError("Report must contain a valid Item object.")
        self._item = value

    @property
    def location(self):
        """Return the campus location of the report."""
        return self._location

    @location.setter
    def location(self, value):
        """Validate that the location is a non-empty string."""
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Location cannot be empty.")
        self._location = value.strip()

    @property
    def reporter(self):
        """Return the User object who submitted the report."""
        return self._reporter

    @reporter.setter
    def reporter(self, value):
        """Validate that reporter is a User object or None."""
        if value is not None and not isinstance(value, User):
            raise ValueError("Reporter must be a User object or None.")
        self._reporter = value

    @property
    def date(self):
        """Return the report date."""
        return self._date

    @date.setter
    def date(self, value):
        """Validate that the date is a non-empty string."""
        if not isinstance(value, str) or not value.strip():
            self._date = "-"
        else:
            self._date = value.strip()

    @property
    def status(self):
        """Return the current report status ('OPEN' or 'RESOLVED')."""
        return self._status

    def update_status(self, new_status):
        """Validate and update the report status to 'OPEN' or 'RESOLVED'."""
        if not isinstance(new_status, str) or not new_status.strip():
            raise ValueError("Status cannot be empty.")
        cleaned = new_status.strip().upper()
        if cleaned not in self.VALID_STATUSES:
            raise ValueError(
                f"Invalid status '{new_status}'. Choose from {self.VALID_STATUSES}."
            )
        self._status = cleaned

    @abstractmethod
    def get_report_type(self):
        """Return the report type string (must be implemented by subclasses)."""
        pass

    @abstractmethod
    def show_details(self):
        """Display formatted details of the report (must be implemented by subclasses)."""
        pass

    def __str__(self):
        """Return a concise one-line summary of the report."""
        return f"[{self._report_id}] ({self.get_report_type()} - {self._status}) {self._item.name} at {self._location}"


class LostReport(Report):
    """Subclass representing a report for an item lost on campus."""

    def __init__(
        self,
        report_id,
        item,
        location,
        reporter=None,
        date="-",
        last_seen_note="-"
    ):
        """Initialize a lost report using the base Report constructor."""
        super().__init__(report_id, item, location, reporter, date)
        self.last_seen_note = last_seen_note

    @property
    def last_seen_note(self):
        """Return additional notes about when the item was last seen."""
        return self._last_seen_note

    @last_seen_note.setter
    def last_seen_note(self, value):
        """Validate and store the last seen note."""
        if not isinstance(value, str) or not value.strip():
            self._last_seen_note = "-"
        else:
            self._last_seen_note = value.strip()

    def get_report_type(self):
        """Return 'LOST' as the report type."""
        return "LOST"

    def show_details(self):
        """Print formatted details specific to a lost item report."""
        reporter_info = self.reporter if self.reporter is not None else "Anonymous"
        print(f"Report ID      : {self.report_id}")
        print(f"Type           : {self.get_report_type()}")
        print(f"Status         : {self.status}")
        print(f"Item           : {self.item}")
        print(f"Lost Location  : {self.location} (Date: {self.date})")
        print(f"Last Seen Note : {self.last_seen_note}")
        print(f"Reported By    : {reporter_info}")


class FoundReport(Report):
    """Subclass representing a report for a found item on campus."""

    def __init__(
        self,
        report_id,
        item,
        location,
        reporter=None,
        date="-",
        pickup_point="Campus Security Desk"
    ):
        """Initialize a found report using the base Report constructor."""
        super().__init__(report_id, item, location, reporter, date)
        self.pickup_point = pickup_point

    @property
    def pickup_point(self):
        """Return the location where the owner can claim the found item."""
        return self._pickup_point

    @pickup_point.setter
    def pickup_point(self, value):
        """Validate that the pickup point is a non-empty string."""
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Pickup point cannot be empty.")
        self._pickup_point = value.strip()

    def get_report_type(self):
        """Return 'FOUND' as the report type."""
        return "FOUND"

    def show_details(self):
        """Print formatted details specific to a found item report."""
        reporter_info = self.reporter if self.reporter is not None else "Anonymous"
        print(f"Report ID      : {self.report_id}")
        print(f"Type           : {self.get_report_type()}")
        print(f"Status         : {self.status}")
        print(f"Item           : {self.item}")
        print(f"Found Location : {self.location} (Date: {self.date})")
        print(f"Pickup Point   : {self.pickup_point}")
        print(f"Reported By    : {reporter_info}")
```

## Bagian 3: Class `LostFoundSystem` dan Menu Console (Layyis)
Bagian ini mengelola daftar laporan, pencarian berdasarkan kata kunci/kategori, pembaruan status, penyimpanan data, serta antarmuka menu konsol interaktif dengan penanganan *error* (`try-except`).

```python
# ==========================================================
# CELL 4 - BAGIAN 3: LostFoundSystem & Console Menu (Layyis)
# ==========================================================
class LostFoundSystem:
    """Manage lost and found reports on campus."""

    # 1. Initialize system
    def __init__(self):
        self._reports = []
        self._counter = 1

    # 2. Generate unique report ID
    def generate_report_id(self):
        while True:
            report_id = f"RPT-{self._counter:03d}"
            self._counter += 1

            if self.find_by_id(report_id) is None:
                return report_id

    # 3. Find a report by ID
    def find_by_id(self, report_id):
        if not isinstance(report_id, str) or not report_id.strip():
            return None

        target_id = report_id.strip().upper()

        for report in self._reports:
            if report.report_id == target_id:
                return report

        return None

    # 4. Add a report
    def add_report(self, report):
        if not isinstance(report, Report):
            raise TypeError(
                "Report must be a Report, LostReport, or FoundReport object."
            )

        if self.find_by_id(report.report_id) is not None:
            raise ValueError("Report ID already exists.")

        self._reports.append(report)
        print(f"Report {report.report_id} added successfully.")

    # 5. Display all reports
    def show_all_reports(self):
        if not self._reports:
            print("No reports available.")
            return

        print("\n========== ALL REPORTS ==========")

        for report in self._reports:
            report.show_details()
            print("-" * 35)

    # 6. Search reports by keyword
    def search_reports(self, query):
        if not isinstance(query, str) or not query.strip():
            print("Search keyword cannot be empty.")
            return []

        keyword = query.strip().lower()
        results = []

        for report in self._reports:
            item = report.item

            searchable_fields = [
                report.report_id,
                report.get_report_type(),
                report.status,
                item.name,
                item.category,
                item.description,
                report.location,
                report.date,
            ]

            # Include reporter information when available
            if report.reporter is not None:
                searchable_fields.extend([
                    report.reporter.user_id,
                    report.reporter.name,
                    report.reporter.contact,
                ])

            # Include subclass-specific information
            if isinstance(report, LostReport):
                searchable_fields.append(report.last_seen_note)

            elif isinstance(report, FoundReport):
                searchable_fields.append(report.pickup_point)

            searchable_text = " ".join(
                str(field) for field in searchable_fields
                if field is not None
            ).lower()

            if keyword in searchable_text:
                results.append(report)

        if not results:
            print("No matching reports found.")
            return []

        print(f"\nFound {len(results)} report(s):")

        for report in results:
            report.show_details()
            print("-" * 35)

        return results

    # 7. Update report status
    def update_report_status(self, report_id, new_status):
        report = self.find_by_id(report_id)

        if report is None:
            print(f"Report with ID {report_id} was not found.")
            return False

        try:
            report.update_status(new_status)

        except (ValueError, TypeError) as error:
            print(f"Unable to update status: {error}")
            return False

        print(
            f"Report {report.report_id} "
            f"status updated to {report.status}."
        )
        return True


# CREATE REPORT
def create_report(system, report_type):
    report_type = str(report_type).strip().upper()

    if report_type not in ("LOST", "FOUND"):
        print("Invalid report type. Use LOST or FOUND.")
        return

    print(f"\n====== CREATE {report_type} REPORT ======")

    try:
        # 1. Input item information
        item_name = input("Item name: ").strip()

        print("\nCategories:")
        print("1. Electronics")
        print("2. Documents")
        print("3. Personal")
        print("4. Others")

        category_options = {
            "1": "Electronics",
            "2": "Documents",
            "3": "Personal",
            "4": "Others",
        }

        category_choice = input("Choose category (1-4): ").strip()
        category = category_options.get(category_choice)

        if category is None:
            print("Invalid category.")
            return

        description = input("Item description: ").strip()
        location = input("Location: ").strip()

        # 2. Create Item object
        item = Item(item_name, category, description)

        # 3. Input optional reporter information
        print("\nReporter information")
        print("Leave all three fields blank for an anonymous report.")

        user_id = input("User ID / NIM: ").strip()
        user_name = input("Reporter name: ").strip()
        contact = input("Contact information: ").strip()

        reporter_fields = [user_id, user_name, contact]

        if all(not field for field in reporter_fields):
            reporter = None

        elif all(field for field in reporter_fields):
            # Create User object
            reporter = User(user_id, user_name, contact)

        else:
            print(
                "Please complete all reporter fields "
                "or leave all three blank."
            )
            return

        # 4. Input report-specific information
        if report_type == "LOST":
            last_seen_note = input(
                "Last seen note (optional): "
            ).strip() or "-"

        else:
            pickup_point = input(
                "Pickup point (press Enter for Campus Security Desk): "
            ).strip() or "Campus Security Desk"

        # 5. Generate report ID after input validation
        report_id = system.generate_report_id()

        # 6. Create the appropriate report
        if report_type == "LOST":
            report = LostReport(
                report_id=report_id,
                item=item,
                location=location,
                reporter=reporter,
                last_seen_note=last_seen_note,
            )

        else:
            report = FoundReport(
                report_id=report_id,
                item=item,
                location=location,
                reporter=reporter,
                pickup_point=pickup_point,
            )

        # 7. Add report to the system
        system.add_report(report)

    except (ValueError, TypeError) as error:
        print(f"Unable to create report: {error}")


# CONSOLE MENU
def run_menu():
    system = LostFoundSystem()

    while True:
        print("\n====== CAMPUS LOST & FOUND ======")
        print("1. Create Lost Report")
        print("2. Create Found Report")
        print("3. Show All Reports")
        print("4. Search Reports")
        print("5. Update Report Status")
        print("0. Exit")

        choice = input("Choose menu: ").strip()

        if choice == "1":
            create_report(system, "LOST")

        elif choice == "2":
            create_report(system, "FOUND")

        elif choice == "3":
            system.show_all_reports()

        elif choice == "4":
            keyword = input("Enter keyword: ")
            system.search_reports(keyword)

        elif choice == "5":
            report_id = input("Enter report ID: ").strip()
            new_status = input(
                "Enter new status (OPEN/RESOLVED): "
            ).strip()

            system.update_report_status(report_id, new_status)

        elif choice == "0":
            print("Thank you for using Campus Lost & Found.")
            break

        else:
            print("Invalid menu choice. Please try again.")
```

## Bagian 4: Pengujian Program / Testing (Normal, Edge, dan Invalid Input)
Bagian ini menguji seluruh class dan metode menggunakan skenario input normal, kasus batas (*edge cases*), dan input tidak valid (*invalid input*).

```python
# ==========================================================
# CELL 5 - TESTING BERSAMA (Normal, Edge, & Invalid Input)
# ==========================================================

# --- NORMAL INPUT SCENARIOS ---
def test_valid_item():
    """TC-01: Valid Item creation & properties."""
    item = Item("Dompet Kulit", "Personal", "Warna cokelat berisi kartu")
    assert item.name == "Dompet Kulit"
    assert item.category == "Personal"


def test_valid_user():
    """TC-02: Valid User creation & string representation."""
    user = User("26120300007", "Evan Adibrata", "evan@campus.ac.id")
    assert user.user_id == "26120300007"
    assert "Evan" in str(user)


def test_valid_lost_report():
    """TC-03: Valid LostReport creation with inheritance & User."""
    item = Item("Laptop ASUS", "Electronics", "Warna abu-abu ada stiker")
    user = User("26120300005", "Nailla", "nailla@campus.ac.id")
    lost = LostReport("RPT-101", item, "Perpustakaan Lt. 2", user, "2026-10-28", "Meja 5")
    assert lost.get_report_type() == "LOST"
    assert lost.status == "OPEN"
    assert lost.last_seen_note == "Meja 5"


def test_valid_found_report_anonymous():
    """TC-04: Valid FoundReport creation (Anonymous reporter)."""
    item = Item("KTM Cakrawala", "Documents", "NIM 26120300004")
    found = FoundReport("RPT-102", item, "Kantin Utama", None, "2026-10-28", "Pos Satpam Gedung A")
    assert found.get_report_type() == "FOUND"
    assert found.pickup_point == "Pos Satpam Gedung A"
    assert found.reporter is None


def test_update_status_resolved():
    """TC-05: Update report status to RESOLVED."""
    sys_test = LostFoundSystem()
    item1 = Item("Tumbler Hijau", "Personal", "Merk Lock&Lock")
    r1 = LostReport("RPT-001", item1, "Lab Komputer", None)
    sys_test.add_report(r1)
    res = sys_test.update_report_status("RPT-001", "RESOLVED")
    assert res is True
    assert sys_test.find_by_id("RPT-001").status == "RESOLVED"


def test_search_by_keyword():
    """TC-06: Search report by item keyword."""
    sys_test = LostFoundSystem()
    item1 = Item("Powerbank Anker", "Electronics", "Warna hitam 10000mAh")
    r1 = FoundReport("RPT-001", item1, "Ruang B201", None)
    sys_test.add_report(r1)
    matches = sys_test.search_reports("Anker")
    assert len(matches) == 1
    assert matches[0].report_id == "RPT-001"


# --- EDGE CASE SCENARIOS ---
def test_whitespace_and_capitalization():
    """TC-07: Automatic whitespace stripping & category capitalization."""
    item = Item("   Payung Lipat   ", "personal", "   Warna biru dongker   ")
    assert item.name == "Payung Lipat"
    assert item.category == "Personal"
    assert item.description == "Warna biru dongker"


def test_case_insensitive_matching():
    """TC-08: Case-insensitive keyword matching (Item.matches)."""
    item = Item("Flashdisk Sandisk", "Electronics", "32GB warna merah")
    assert item.matches("sandisk") is True
    assert item.matches("ELECTRONICS") is True
    assert item.matches("FLASHDISK") is True
    assert item.matches("tas") is False


def test_case_insensitive_id_and_defaults():
    """TC-09: Case-insensitive ID search & default value fallback."""
    sys_test = LostFoundSystem()
    item1 = Item("Kacamata Minus", "Personal", "Frame hitam")
    r1 = LostReport("rpt-001", item1, "Musholla", date="")
    sys_test.add_report(r1)
    assert sys_test.find_by_id("RPT-001") is not None
    assert r1.date == "-"
    assert r1.last_seen_note == "-"


def test_search_no_result():
    """TC-10: Search on system with no matching results."""
    sys_test = LostFoundSystem()
    matches = sys_test.search_reports("barang_tidak_ada")
    assert matches == []


# --- INVALID INPUT SCENARIOS ---
def test_reject_empty_item_name():
    """TC-11: Reject empty Item name with ValueError."""
    try:
        Item("", "Personal", "Deskripsi valid")
        assert False, "Should raise ValueError on empty name"
    except ValueError:
        pass


def test_reject_invalid_category():
    """TC-12: Reject disallowed Item category with ValueError."""
    try:
        Item("Jaket Kampus", "Fashion", "Warna navy")
        assert False, "Should raise ValueError on invalid category"
    except ValueError:
        pass


def test_abstract_report_not_instantiable():
    """TC-13: Prevent direct instantiation of abstract Report class."""
    try:
        item = Item("Buku Tulis", "Others", "Garis-garis")
        Report("RPT-999", item, "Kantin")
        assert False, "Should raise TypeError when instantiating abstract Report"
    except TypeError:
        pass


def test_reject_invalid_status():
    """TC-14: Reject invalid status value with ValueError."""
    item = Item("Kunci Motor", "Personal", "Gantungan boneka")
    lost = LostReport("RPT-005", item, "Parkiran Barat")
    try:
        lost.update_status("CLOSED_NOW")
        assert False, "Should raise ValueError on invalid status"
    except ValueError:
        pass


# (test ID, category, description, test function)
TEST_CASES = [
    ("TC-01", "Normal ", "Valid Item creation & properties", test_valid_item),
    ("TC-02", "Normal ", "Valid User creation & string representation", test_valid_user),
    ("TC-03", "Normal ", "Valid LostReport creation with inheritance & User", test_valid_lost_report),
    ("TC-04", "Normal ", "Valid FoundReport creation (Anonymous reporter)", test_valid_found_report_anonymous),
    ("TC-05", "Normal ", "Update report status to RESOLVED", test_update_status_resolved),
    ("TC-06", "Normal ", "Search report by item keyword", test_search_by_keyword),
    ("TC-07", "Edge   ", "Automatic whitespace stripping & category capitalization", test_whitespace_and_capitalization),
    ("TC-08", "Edge   ", "Case-insensitive keyword matching (Item.matches)", test_case_insensitive_matching),
    ("TC-09", "Edge   ", "Case-insensitive ID search & default value fallback", test_case_insensitive_id_and_defaults),
    ("TC-10", "Edge   ", "Search on system with no matching results", test_search_no_result),
    ("TC-11", "Invalid", "Reject empty Item name with ValueError", test_reject_empty_item_name),
    ("TC-12", "Invalid", "Reject disallowed Item category with ValueError", test_reject_invalid_category),
    ("TC-13", "Invalid", "Prevent direct instantiation of abstract Report class", test_abstract_report_not_instantiable),
    ("TC-14", "Invalid", "Reject invalid status value with ValueError", test_reject_invalid_status),
]


def run_test_case(test_id, category, description, assertion_fn):
    """Run one test case, print PASS/FAIL, and return True if it passed."""
    try:
        assertion_fn()
        print(f"[PASS] {test_id} ({category}): {description}")
        return True
    except Exception as error:
        print(f"[FAIL] {test_id} ({category}): {description} -> Error: {error}")
        return False


def print_test_summary(passed, failed, total):
    """Print the total, passed, failed, and pass rate of the test suite."""
    pass_rate = (passed / total) * 100
    print("=" * 60)
    print("                AUTOMATED TEST SUITE SUMMARY")
    print("=" * 60)
    print(f"  Total Test Cases : {total}")
    print(f"  Passed           : {passed}")
    print(f"  Failed           : {failed}")
    print(f"  Pass Rate        : {pass_rate:.1f}%")
    if passed == total:
        print("  Evaluation       : EXCELLENT - ALL UNIT TESTS PASSED (100%)")
    else:
        print("  Evaluation       : SOME TESTS FAILED - PLEASE REVIEW")
    print("=" * 60)


def run_automated_tests():
    """Execute automated test suite covering Normal, Edge, and Invalid input scenarios."""
    print("=" * 60)
    print("      CAMPUS LOST & FOUND - AUTOMATED TEST SUITE")
    print("=" * 60)
    passed = 0
    for test_id, category, description, assertion_fn in TEST_CASES:
        if run_test_case(test_id, category, description, assertion_fn):
            passed += 1
    total = len(TEST_CASES)
    print_test_summary(passed, total - passed, total)


run_automated_tests()
```
